```python
import pandas as pd
import numpy  as nd 

df_vendas = pd.read_csv("vendas_tech.csv", sep =",")
display(df_vendas)
df_gerentes = pd.read_excel("gerentes_lojas.xlsx")
display(df_gerentes)
```

```python
# tratamento de dados
# display(df_vendas["Loja"])
# display(df_vendas[["Loja", "Cliente"]])


#colunas 
df_analise = df_vendas.drop(columns = ["Data_Base"]) 
# usamos o drop para apagar/deletar uma coluna que aqui foi a "Data_Base"


# nulos  
# df_analise = df_analise.dropna()  
# # dropna excluir todas as linhas que têm um valor vazio 
#fillna preenche os valores vazios 
df_analise["Loja"] = df_analise["Loja"].fillna("Onlinne")  


# tipos de dados
#datetime é transformar em datas  
df_analise["DataFormatada"] = pd.to_datetime(df_analise["Data"], format="%Y-%m-%d") 
# padronização 
# str disponibiliza as funções de texto dentro do Python # strip que remove espaços duplicados
# title é usado para padronizar as primeiras letras de cada palavra em maiúsculo e o resto em minusclo
df_analise["Loja"] = df_analise["Loja"].str.strip()
df_analise["Loja"] = df_analise["Loja"].str.title()  


# duplicatas remover linhas iguais 
df_analise = df_analise.drop_duplicates(subset=["ID_Pedido"]) 
# duplicates()remove os duplicados 
# subset= ele permite selecionar uma lista de colunas específica que quiser para sair os duplicados.
# keep= é manter, você vai definir se vai manter o primeiro (first) ou o último (last) valor duplicado, se não passar nada ele mantém sempre o PRIMEIRO VALOR.

display(df_analise)
display(df_analise.info())
```

```python
# criar novas colunas 


# faturamento 
df_analise["Faturamento"] = df_analise["Qtd"] * df_analise["Preco_Unitario"]

# regiao
# unique são os valores únicos dentro da tabela
display(df_analise["Loja"].unique()) 

def definir_regiao(loja):
    dic_regioes = {
        "São Paulo": "Sudeste", 
        "Belo Horizonte": "Sudeste",   
        "Online": "Online", 
        "Rio de Janeiro": "Sudeste",
        "Salvador": "Nordeste",         
        "Recife": "Nordeste",      
        "Curitiba": "Sul",   
        "Porto Alegre": "Sul"          
    }  
    
    return dic_regioes.get(loja)

df_analise["Região"] = df_analise["Loja"].map(definir_regiao)



# forma de venda 
import numpy as np
df_analise["Forma_de_Venda"] = np.where(df_analise["Loja"]== "Onlinne", "Online", "Presencial")  
# np.where é uma condição 


display(df_analise)
```

```python
# analisar usando filtros/ como filtrar a base de dados

df_analise = df_analise.sort_values(by = "Data")  # ordenando a tabela em data 

# .loc localiza por nome da coluna e por nome da linha que é o índice
# Exemplo usando o .loc : id do pedido
id_pedido = 4 
loja = df_analise.loc[df_analise["ID_Pedido"]==4,"Loja"]         # .loc[indice, coluna]
produto = df_analise.loc[df_analise["ID_Pedido"]==4, "Produto"]  # .loc[indice, coluna]      
cliente = df_analise.loc[df_analise["ID_Pedido"]==4, "Cliente"]  # .loc[indice, coluna]
print(loja,produto, cliente)            # coloca o print paa mostrar as 3 informações de cima 


# .iloc localiza por posição 
# Exemplo usando .iloc 
id_pedido = 4 
loja = df_analise.iloc[3,2]        #.iloc[linha, coluna]
produto = df_analise.iloc[3,3]        # .iloc[linha, coluna]
cliente = df_analise.iloc [3,6]       # .iloc[linha, coluna]
print(loja, produto, cliente)

# condicional pode passar por regra de linha ou uma linha inteira
df_id_pedido_4 = df_analise[df_analise["ID_Pedido"]==4] # aqui passou a condição 


# exportar pedaços da base 
df_vendas_sp = df_analise[df_analise["Loja"]=="São Paulo"]
display(df_vendas_sp)

# exportar as vendas de 2024  
df_analise["Data"] = pd.to_datetime(df_analise["Data"])  
df_vendas_2024 = df_analise[df_analise["Data"].dt.year==2024]
display(df_vendas_2024)  # mostra as vendas apenas de 2024

# duplas condições 
condicao = df_analise["Produto"]=="Cabo HDMI"
condicao2 = df_analise["Região"]=="Sul"
df_vendas_HDMI_SUL = df_analise[condicao & condicao2]
display(df_vendas_HDMI_SUL)
```

```python
# analisar por agrupamento
display(df_analise)

analise_lojas = df_analise[["Loja","Faturamento"]].groupby("Loja").sum()  # groupby agrupa uma coluna do banco de dados
display(analise_lojas)

# ranking de produtos que mais venderam no online
df_vendas_onlinne = df_analise[df_analise["Loja"]=="Onlinne"]
analise_produtos_onlinne = df_vendas_onlinne[["Produto", "Qtd"]].groupby("Produto").sum()
analise_produtos_onlinne = analise_produtos_onlinne.sort_values(by="Qtd", ascending=False)
analise_produtos_onlinne = analise_produtos_onlinne.rename(columns={"Qtd": "Vendas Totais"}) # alterando o nome da coluna Qtd para Vendas Totais 
display(analise_produtos_onlinne)

```
