```python
import pandas as pd
df = pd.read_excel("reshaping.xlsx")

df
```

```python
# RESHAPING: É pegar os dados e mudar a forma como eles estão organizados (linhas ↔ colunas) para facilitar análise.

df.index
```

```python
df.columns
```

```python
# PIVOT()
# Uma operação PIVOT em um dataframe é uma operação que reorganiza os dados de um dataframe, alterando a forma dos dados. 

df_pivot = df.pivot(index=["Estudante"],columns= ["Categoria","Semana", "Tópico"], values="Nota")
```

```python
df_pivot
```

```python
df_pivot.index
```

```python
df_pivot.columns
```

```python
# STACK () empilha colunas e transforma em linhas.

df_pivot.stack(level=0)
```

```python
df_pivot.stack(level=0).index
```

```python
df_pivot.stack(level=0).columns
```

```python
df_pivot.stack(level=0).columns     
```

```python
# UNSTACK () Faz o contrário do stack(), transforma índices em colunas.
df_pivot.stack(level=0).unstack(level=0)
```

```python
# PIVOT TABLE () Parecido com Pivot, mas aceita valores duplicados porque faz agregação.
# É uma operação de reorganização e agregação que toma uma tabela de dados e transforma em uma tabela de resumo. 
# A operação é útil para resumir e reorganizar dados.
# Se você não passa alguma função de agregação ele faz automaticamente a média
df_pivot = df.pivot(index=["Estudante"],columns= ["Categoria","Semana", "Tópico"], values="Nota")

df_pivot
```

```python
df.pivot_table(index=["Estudante"],columns= ["Semana", "Tópico"], values="Nota")
```

```python
# Passando a função de agregação (aggfunc) de contagem (count)
# aggfunc é função de agregação 
df.pivot_table(index=["Estudante"],columns= ["Semana", "Tópico"], values="Nota", aggfunc="count")
```

```python
df.pivot_table(index=["Estudante"],columns= ["Semana", "Tópico"], values="Nota", aggfunc="sum")
```

```python
df.pivot_table(index=["Estudante"],columns= ["Semana", "Tópico"], values="Nota", aggfunc="max")
```

```python
df.pivot_table(index=["Estudante"],columns= ["Semana", "Tópico"], values="Nota", aggfunc="min")
```

```python
df.groupby(["Estudante", "Semana","Tópico"])["Nota"].mean()
```

```python
df.groupby(["Estudante", "Semana","Tópico"])["Nota"].mean().index
```

```python
df.groupby(["Estudante", "Semana","Tópico"])["Nota"].mean().unstack(level=[1,2]),
```
