```python
# USANDO O QUERY FICA MAIS FÁCIL A LEITURA DOS CÓDIGOS

import pandas as pd

df = pd.read_csv("cereal.csv")

df.head()
```

```python
# Condição Simples 
df[df["rating"] > 70]
```

```python
# USANDO O QUERY
df.query("rating > 70 ")
```

```python
df.query("rating > 70 ")[["name","rating"]]
```

```python
df[df["name"] == "Almond Deligth"]
```

```python
df.query("name == 'Almond Deligth'")
```

```python
df[df["mfr"] =="N"]
```

```python
# OPERADOR AND no Pandas é o &
df[(df["mfr"]=="N") & (df["type"]=="H")]
```

```python
# Mais fácil usando o QUERY usando AND
df.query("mfr=='N' and type =='H'")
```

```python
# No pandas o AND é representado pelo & 
# No pandas o NOT e representado pelo ~     
df[(df["mfr"] == "N") & ~(df["type"] == "H")]
```

```python
df.query("mfr == 'N'and not type == 'H'")
```

```python
df.query("mfr == 'N' or type =='H'")
```

```python
# No pandas o OR é representado | pela barra 

df[(df["mfr"] == "N") | (df["type"] == "H")]
```

```python
df[df["mfr"].isin(["R", "N", "A"])]
```

```python
df.query("mfr in ['R','N','A']")
```

```python
# VARIÁVEIS 

empresas_de_interesse = ["R", "N", "A"]
df[df["mfr"].isin(empresas_de_interesse)]
```

```python
# PARA PASSAR VARIÁVEIS NO MÉTODO QUERY USAMOS O @
df.query("mfr in @empresas_de_interesse"    )
```

```python
# CONDIÇÕES COM OPERAÇÕES MATEMÁTICAS

df.query("carbo + sugars < 20")
```
