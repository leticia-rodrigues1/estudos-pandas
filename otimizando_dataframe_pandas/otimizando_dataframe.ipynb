```python
df_otimizado = df.copy()
df_otimizado.info()
```

```python
df_otimizado["data_transacao"] = pd.to_datetime(df_otimizado["data_transacao"],format="%d/%m/%Y")

df_otimizado.info()
```

```python
df_otimizado.head()
```

```python
# informações por ano/year
df_otimizado["data_transacao"].dt.year
```

```python
# informações por mês/month
df_otimizado["data_transacao"].dt.month
```

```python
# informações por dia/ day
df_otimizado["data_transacao"].dt.day
```

```python
# astype serve para alterar o tipo de dado de uma coluna.
# category é um tipo de dado do Pandas usado para colunas com valores repetidos / categorias fixas.

# Usando aqui o category ele mostra pra gente em textos mas armazena os dados como números, com isso diminuindo o armazenamento interno
df_otimizado["categoria_produto"]=df_otimizado["categoria_produto"].astype("category")
df_otimizado["genero"]=df_otimizado["genero"].astype("category")

df_otimizado.info()
```

```python
# sys ele permite interagir com o próprio sistema Python.
# Se você viu sys junto com Pandas, normalmente era para medir memória.

import sys 

sys.maxsize
```

```python
sys.float_info.max
```

```python
colunas_float = df_otimizado.select_dtypes(include="float64").columns
colunas_int = df_otimizado.select_dtypes(include="int64").columns

colunas_float,colunas_int
```

```python
df_otimizado[colunas_float] = df_otimizado[colunas_float].apply(pd.to_numeric,downcast="float")
df_otimizado[colunas_int] = df_otimizado[colunas_int].apply(pd.to_numeric,downcast="integer")
```
