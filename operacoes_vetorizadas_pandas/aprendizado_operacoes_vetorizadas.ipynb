```python
import pandas as pd

base = pd.read_csv("dados_transacoes (2).csv")

df = pd.read_csv("dados_transacoes (2).csv")

df.head()
```

```python
df.info()
```

```python


# loop for 

%timeit

for i in range(len(df)):
    df.loc[i,"total_for"] = df.loc[i, "quantidade"] * df.loc[i, "preco"]

```

```python
df.head()
```

```python
df["total_apply"] = df.apply(
    lambda linha: linha["quantidade"] * linha["preco"],axis=1
)

df.head()
```

```python
# operação vetorizada 

df["total_vetorizado"] = df["quantidade"] * df["preco"]
```

```python
descontos_categorias = {
    "Eletronicos": 0.1,
    "Alimentos": 0.05,
    "Roupas":0.15,
    "Casa & Cozinha": 0.2,
    "Livros":0.25,
}
```
