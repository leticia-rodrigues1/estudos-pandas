O `pandas.crosstab` é uma função que permite calcular tabelas de contingência (também conhecidas como tabelas cruzadas) que resumem a frequência de combinações de valores em duas ou mais colunas categóricas. Essa função é útil para explorar a relação entre diferentes categorias de dados, facilitando a visualização de como as categorias se distribuem e interagem.

Além da frequência (contagem), outras funções de agregação podem ser solicitadas.

Documentação: https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.crosstab.html

Use o `pandas.crosstab` quando:

1. **Explorar relações categóricas**: Você quer explorar a relação entre duas ou mais variáveis categóricas, como a distribuição de vendas por região e produto, ou a frequência de respostas por faixa etária e gênero.
2. **Análise de dados**: Você precisa criar uma tabela de frequência para entender melhor como as diferentes categorias se comportam em conjunto.
3. **Preparação de dados para visualização**: Você está preparando dados para visualizações como gráficos de barras empilhadas, gráficos de calor ou outros tipos de gráficos que se beneficiam de tabelas de contigência.


Contexto:

Uma empresa que vende produtos em várias regiões. Cada produto pode ser vendido em várias regiões, e você tem um registro das vendas de cada produto em cada região para um determinado período. Você quer analisar a distribuição dessas vendas.

```python
import pandas as pd

BASE = "./dados/produtos_regioes.csv"

df = pd.read_csv(BASE)

df
```

```python
# Calculando a tabela de contingência com contagens
pd.crosstab(df["regiao"], df["produto"])
```

```python
pd.crosstab(df["regiao"], df["produto"], margins=True, margins_name="Vendas")
```

## Entendendo o normalize

```python
pd.crosstab(
    df["regiao"], df["produto"], normalize=True, margins=True, margins_name="Total"
)
```

```python
pd.crosstab(
    df["regiao"], df["produto"], normalize=True, margins=True, margins_name="Total"
).mul(100).round(1)
```

```python
pd.crosstab(df["regiao"], df["produto"], margins=True, margins_name="Vendas")
```

```python
pd.crosstab(
    df["regiao"], df["produto"], normalize="index", margins=True, margins_name="Total"
).mul(100).round(1)
```

```python
pd.crosstab(df["regiao"], df["produto"], margins=True, margins_name="Vendas")
```

```python
pd.crosstab(
    df["regiao"], df["produto"], normalize="columns", margins=True, margins_name="Total"
).mul(100).round(1)
```

## Usando outras funções de agregação

```python
# agora vamos calcular a tabela de contingência com a receita total
pd.crosstab(
    df["regiao"],
    df["produto"],
    values=df["receita"],
    aggfunc="sum",
    margins=True,
    margins_name="Receita Total",
)
```

```python
# agora vamos calcular a tabela de contingência com a receita média
pd.crosstab(
    df["regiao"],
    df["produto"],
    values=df["receita"],
    aggfunc="mean",
    margins=True,
    margins_name="Receita Média",
).round(2)
```

```python
pd.crosstab(
    df["regiao"],
    df["produto"],
    values=df["receita"],
    aggfunc="sum",
    normalize=True,
).mul(100).round(1)
```

```python
pd.crosstab(
    df["regiao"],
    df["produto"],
    values=df["receita"],
    aggfunc="sum",
    normalize="index",
).mul(100).round(1)
```

```python
pd.crosstab(
    df["regiao"],
    df["produto"],
    values=df["receita"],
    aggfunc="sum",
    normalize="columns",
).mul(100).round(1)
```

## Comparando com groupby

```python
pd.crosstab(df["regiao"], df["produto"])
```

```python
# obtendo a contagem igual ao crosstab mas com a função groupby
df.groupby(["regiao", "produto"]).count().rename(columns={"receita": "vendas"})
```

```python
df.groupby(["regiao", "produto"]).count().unstack().fillna(0).rename(
    columns={"receita": "vendas"}
)
```

```python
pd.crosstab(
    df["regiao"],
    df["produto"],
    values=df["receita"],
    aggfunc="sum",
    margins=True,
    margins_name="Receita Total",
)
```

```python
# obtendo a soma igual ao crosstab mas com a função groupby
df.groupby(["regiao", "produto"]).sum().rename(columns={"receita": "receita_total"})
```

```python
df.groupby(["regiao", "produto"]).sum().unstack().fillna(0).rename(
    columns={"receita": "receita_total"}
)
```

### Comparação entre `crosstab` e `groupby`

Vantagens do `crosstab`:

- Simplicidade: `crosstab` é mais direto e simples de usar para criar tabelas de contingência.
- Automático: Ele lida automaticamente com os valores ausentes e adiciona margens (totais) de maneira conveniente.

Desvantagens do `crosstab`:

- Flexibilidade Limitada: Menos flexível que `groupby` para aplicar múltiplas agregações diferentes ou manipular dados de maneiras mais complexas.
- Personalização: Menos opções para personalizar a tabela resultante.

Vantagens do `groupby`:

- Flexibilidade: Muito mais flexível, permitindo aplicar múltiplas funções de agregação e manipular dados de várias maneiras.
- Versatilidade: Pode ser usado para uma ampla variedade de operações de agrupamento e agregação, não apenas para tabelas de contingência.

Desvantagens do `groupby`:

- Complexidade: Pode ser mais complexo e requer mais código para realizar operações simples que `crosstab` faz facilmente.
- Margens Manualmente: Adicionar margens (totais) requer operações adicionais, o que pode ser menos conveniente.

## Comparando com pivot_table

```python
pd.crosstab(df["regiao"], df["produto"])
```

```python
# o mesmo resultado pode ser obtido com a função pivot_table
df.pivot_table(
    index="regiao", columns="produto", values="receita", aggfunc="count", fill_value=0
)
```

```python
pd.crosstab(
    df["regiao"],
    df["produto"],
    values=df["receita"],
    aggfunc="sum",
    margins=True,
    margins_name="Receita Total",
)
```

```python
# o mesmo resultado pode ser obtido com a função pivot_table
df.pivot_table(
    index="regiao",
    columns="produto",
    values="receita",
    aggfunc="sum",
    fill_value=0,
    margins=True,
    margins_name="Receita Total",
)
```

### Comparação entre crosstab e pivot_table

Vantagens do `crosstab`:

- Simplicidade: `crosstab` é mais direto e simples de usar para criar tabelas de contingência.
- Automático: Ele lida automaticamente com valores ausentes e adiciona margens de maneira conveniente.
- Desempenho: Pode ser mais rápido e eficiente para grandes conjuntos de dados categóricos.

Desvantagens do `crosstab`:

- Flexibilidade Limitada: Menos flexível que `pivot_table` para aplicar múltiplas agregações diferentes ou manipular dados de maneiras mais complexas.
- Personalização: Menos opções para personalizar a tabela resultante.

Vantagens do `pivot_table`:

- Flexibilidade: Muito mais flexível, permitindo aplicar múltiplas funções de agregação e manipular dados de várias maneiras.
- Versatilidade: Pode ser usado para uma ampla variedade de operações de agrupamento e agregação, não apenas para tabelas de contingência.
- Suporte a Vários Agregadores: Pode aplicar múltiplas funções de agregação ao mesmo tempo, como `sum`, `mean`, `count`, etc.

Desvantagens do `pivot_table`:

- Complexidade: Pode ser mais complexo e requer mais código para realizar operações simples que `crosstab` faz facilmente.
