```python
import pandas as pd

df = pd.read_csv("produtos_regioes.csv")

df
```

```python
pd.crosstab(df["regiao"], df["produto"])
```

```python
# margins é adicionar totais automáticos na tabela
# margins são os totais 
pd.crosstab(df["regiao"], df["produto"], margins=True)
pd.crosstab(df["regiao"], df["produto"], margins=True, margins_name="Vendas")
```

```python
# normalize 
# normalizar 
# transforma contagens em porcentagens/proporções

pd.crosstab(df["regiao"], df["produto"],normalize=True, margins=True, margins_name="Total")
```

```python
pd.crosstab(df["regiao"], df["produto"],normalize=True, margins=True, margins_name="Total")
```

```python


pd.crosstab(df["regiao"], df["produto"],normalize=True, margins=True, margins_name="Total")
```

```python
# usamos a multiplicação * vezes 100
pd.crosstab(
    df["regiao"], df["produto"],normalize=True, margins=True, margins_name="Total"
    )*100
```

```python
pd.crosstab(
    df["regiao"], df["produto"],normalize="index", margins=True, margins_name="Total"
    ).mul(100).round(1)
```

```python
pd.crosstab(
    df["regiao"], df["produto"],normalize="columns", margins=True, margins_name="Total"
    ).mul(100).round(1)
```

```python
# USANDO OUTRAS FUNÇÕES DE AGREGAÇÃO 
pd.crosstab(df["regiao"], df["produto"])
pd.crosstab(df["regiao"], df["produto"],values=df["receita"], aggfunc="sum",
           margins=True, margins_name = "Receita Total" )
```

```python
pd.crosstab(
    df["regiao"],
    df["produto"],
    values=df["receita"], 
    aggfunc="sum",
    margins=True, 
    margins_name = "Receita Total" 
    ).round(2)
# ROUND serve para ARREDONDAR números decimais 
```

```python
pd.crosstab(
    df["regiao"],
    df["produto"],
    values=df["receita"], 
    aggfunc="sum",
    normalize=True, 
    ).mul(100).round(1)
# MUL é multiplicação
# ROUND serve para ARREDONDAR números decimais 
```

```python
pd.crosstab(
    df["regiao"],
    df["produto"],
    values=df["receita"], 
    aggfunc="sum",
    normalize="index", 
    ).mul(100).round(1)
# as referêcias aqui são as linhas/ index
# INDEX = índice
# a numeração/identificador das linhas da tabela
```

```python
pd.crosstab(
    df["regiao"],
    df["produto"],
    values=df["receita"], 
    aggfunc="sum",
    normalize="columns", 
    ).mul(100).round(1)

# A referência aqui são as COLUNAS/ COLUMNS
```

```python
# COMPARANDO COM O GROUPBY
pd.crosstab(df["regiao"], df["produto"])
```

```python
# USANDO GROUPBY

# UNSTACK ( )
# unstack = desempilhar
# tirar informações que estão empilhadas nas linhas (índice) e transformar em colunas.
df.groupby(["regiao","produto"]).count().unstack().fillna(0).rename(columns={"receita":"vendas"})
```
