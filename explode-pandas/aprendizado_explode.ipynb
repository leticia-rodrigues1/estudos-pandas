```python
import pandas as pd
dados = pd.read_csv("dados.csv")

display(dados)
```

```python
# EXPLODE = explodir uma lista em várias linhas

dados.info()
```

```python
dados.loc[0,"fases"]
```

```python
type(dados.loc[0,"fases"])
```

```python
for coluna in ["fases", "horas", "taxas_horarias"]:
    remover = ["[", "]", "'", " "]

    for simbolo in remover:
        dados[coluna] = dados[coluna].str.replace(simbolo, "")

    dados[coluna] = dados[coluna].str.split(",")

dados
```

```python
dados.info()
```

```python
dados.loc[0, "fases"]
```

```python
type(dados.loc[0,"fases"])
```

```python
# USANDO O EXPLODE 

dados_expandido = dados.explode(["fases", "horas", "taxas_horarias"])

dados_expandido
```

```python
dados_expandido.info()
```

```python
dados_expandido["data_pedido"] = pd.to_datetime(dados_expandido["data_pedido"], format="%Y-%m")
dados_expandido 
```

```python
dados_expandido.info()
```

```python
for coluna in ["horas", "taxas_horarias"]:
    dados_expandido[coluna] = pd.to_numeric(dados_expandido[coluna])

dados_expandido
```

```python
dados_expandido.info()
```

```python
# Total de horas por fase
dados_expandido.groupby("fases")["horas"].sum()

```

```python
# TOTAL DE HORAS POR FASE POR MÊS

dados_expandido.groupby([dados_expandido["data_pedido"].dt.month, "fases"])["horas"].sum()
```
